```python
import pandas as pd
from sklearn import linear_model
from sklearn.feature_extraction.text import TfidfVectorizer
#加载数据集
df = pd.read_csv('SMSSpamCollection.txt',delimiter='\t',header=None)
#将数据集分为训练数据集
y,X_train = df[0],df[1]
#转化为向量
vectorizer = TfidfVectorizer()
#训练数据集
X = vectorizer.fit_transform(X_train)
```

```python
#利用linear_model构造逻辑回归分类器
clf=linear_model.LogisticRegression(max_iter=200,solver='liblinear')
clf.fit(X,y)
```

```python
#测试集
testX = vectorizer.transform([ 'You are awarded a SiPix Digital Camera! call 09061221061 from landline. Delivery within 28days. T Cs Box177. M221BP. 2yr warranty. 150ppm. 16 . p p£3.99','Please call our customer service representative on 0800 169 6031 between 12am-13pm as you have WON a guaranteed £8000 cash or £5000 prize!'])
#对语句进行预测分类，输出结果
results=clf.predict(testX)
print(results)
```
